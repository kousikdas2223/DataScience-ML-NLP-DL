```python
import matplotlib.pyplot as plt

# Create a simple figure and axis
x_axis = [1,2,3,4,5]
y_axis = [10,20,-25,30,35]
plt.plot(x_axis, y_axis, color="red", linestyle="--", marker="o")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("Simple plot")
plt.show()
```
